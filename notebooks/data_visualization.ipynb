```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df_cleaned = pd.read_csv("data/cleaned_creditcard.csv")

# Distribution of 'Amount' by class
plt.figure(figsize=(10, 6))
sns.histplot(data=df_cleaned, x="Amount", hue="Class", bins=50, kde=True)
plt.title("Transaction Amount Distribution by Class")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Class", data=df_cleaned)
plt.title("Transaction Count by Class")
plt.xlabel("Class (0 = Legit, 1 = Fraud)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Correlation matrix
corr_matrix = df_cleaned.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap="coolwarm", annot=False)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import PCA

# Apply PCA to reduce features to 2 dimensions
X = df_cleaned.drop("Class", axis=1)
y = df_cleaned["Class"]

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Create DataFrame for plotting
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Class"] = y

# Plot PCA
plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Class", alpha=0.5)
plt.title("PCA Projection (2D)")
plt.tight_layout()
plt.show()
```
